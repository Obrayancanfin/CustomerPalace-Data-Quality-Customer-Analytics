```python
df_raw = cleaning_df.copy(deep=True)
df_clean = cleaning_df.copy(deep=True)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)
```

```python
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
nb_doublons = int(df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Lignes dupliquées supprimées :", nb_doublons)
```

```python
num_cols = df_clean.select_dtypes(
    include="number"
).columns

for col in num_cols:
    if df_clean[col].isna().any():
        mediane = df_clean[col].median()
        print(f"{col} : médiane utilisée = {mediane}")
```

```python
for col in num_cols:
    if df_clean[col].isna().any():
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )
```

```python
def detecter_outliers_iqr(data, colonne):
    q1 = data[colonne].quantile(0.25)
    q3 = data[colonne].quantile(0.75)
    iqr = q3 - q1

    borne_inf = q1 - 1.5 * iqr
    borne_sup = q3 + 1.5 * iqr

    masque = (
        (data[colonne] < borne_inf) |
        (data[colonne] > borne_sup)
    )

    return data.loc[masque], borne_inf, borne_sup


# Exemple :
# outliers, borne_inf, borne_sup = detecter_outliers_iqr(
#     df_clean, "montant"
# )
# display(outliers)
```
